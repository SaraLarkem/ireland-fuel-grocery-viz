# Notebook: 01b_Data_Cleaning_Grocery_Items.ipynb

**Goal:** Ingest and standardize the raw Consumer Price Index (CPI) item data from CSO Ireland.

The raw data is correctly formatted but not yet optimized for time-series analysis. This notebook converts string-based dates (e.g., "2016 January") into searchable datetime objects, removes redundant categorical metadata, and prepares the clean file required for Chenrui's subcategory analysis (Notebook 04).

Source: CSO table CPM12 (https://data.cso.ie/table/CPM12), downloaded 21 November 2026 (original file name CPM12.20261121T201144.csv).


```python
import os
import pandas as pd

# Load the raw data from the repo's local data/ folder (works from notebooks/ in VS Code or from the repo root)
DATA_DIR = '../data' if os.path.basename(os.getcwd()) == 'notebooks' else 'data'
df = pd.read_csv(f'{DATA_DIR}/raw_perishable_non_perishable_data.csv')

# Display initial state
print("--- Initial Raw Data ---")
print(df.head(3))
```

### 1. Data Standardization and Type Conversion
To perform rolling averages and lag correlations later in the pipeline, the `Date` column must be a formal Pandas Datetime object rather than plain text. We will also strip out the `STATISTIC Label` and `UNIT` columns, as every row identically represents the "National Average Price" in "Euros", making those columns redundant for programmatic analysis.

```python
# 1. Rename columns for easier access and coding consistency
df = df.rename(columns={
    'STATISTIC Label': 'Statistic',
    'Month': 'Date',
    'Consumer Item': 'Item',
    'UNIT': 'Unit',
    'VALUE': 'Price'
})

# 2. Convert 'Date' string (e.g., "2016 January") to a proper datetime object
df['Date'] = pd.to_datetime(df['Date'], format='%Y %B')

# 3. Drop unnecessary columns
df = df.drop(columns=['Statistic', 'Unit'])

# 4. Sort chronologically and alphabetically by item to ensure a clean index
df = df.sort_values(by=['Date', 'Item']).reset_index(drop=True)

# 5. Basic validation check to confirm cleaning success
print(f"Cleaned dataset range: {df['Date'].min().date()} to {df['Date'].max().date()}")
print(f"Total rows: {len(df)}")
print("\n--- Cleaned Data Sample ---")
print(df.head())
```

### 2. Completeness Check
Notebook 04 averages the item prices for each month, so a missing item in any month would distort that month's average. Before exporting, we confirm there are no missing values, that every month contains the same number of items, and that no item appears twice in the same month.

```python
# 1. Count missing values in each column
print("Missing values:\n", df.isnull().sum())

# 2. Count how many items each month has (every month should have the same number)
print("\nItems per month:\n", df.groupby('Date').size().value_counts())

# 3. Check that no item appears twice in the same month
print("\nDuplicate rows:", df.duplicated(subset=['Date', 'Item']).sum())
```

**Coverage:** January 2016 – October 2026 (130 months, 14 items). This now includes the 2026 Strait of Hormuz period, and runs one month past the fuel data, which ends in September 2026.

### 3. Export to Pipeline
With the dates properly formatted and the columns standardized, this clean dataset is exported to the `data/` directory.
* **(Notebook 04)** will use this output to pivot the items into perishable (meat, dairy, fresh produce) vs. shelf-stable (cereals, bread) groups to analyze differing reactions to fuel spikes.

```python
import os

# Use the repo's data/ folder whether the notebook runs from notebooks/ (VS Code) or the repo root (Colab)
DATA_DIR = '../data' if os.path.basename(os.getcwd()) == 'notebooks' else 'data'

# Ensure the data directory exists before exporting
os.makedirs(DATA_DIR, exist_ok=True)

# Save the cleaned dataset for Notebook 04
export_path = f'{DATA_DIR}/cleaned_grocery_subcategories.csv'
df.to_csv(export_path, index=False)

print(f"Cleaning complete. Data exported to: {export_path}")
```
